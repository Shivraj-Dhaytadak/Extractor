```python
from langchain_ollama.chat_models import ChatOllama
from langchain_google_genai import ChatGoogleGenerativeAI
from langchain_core.prompts import ChatPromptTemplate
from langchain_core.messages import HumanMessage, SystemMessage
from pydantic import BaseModel


class Question(BaseModel):
    question: str
    options: list[str]

class QuestionList(BaseModel):
    questions: list[Question]

json_path = r"C:\VScodeMaster\Inferencing\Extractor\application\Experiment\data\json\apigateway.json"
with open(json_path, 'r', encoding='utf-8') as file:
    data = file.read()
service_name = "apigateway"
llm = ChatGoogleGenerativeAI(
    api_key="AIzaSyDPaPrhk1d9JDjWH-rnE4CdEWUgkWUgO7E",
    model="gemini-2.0-flash-lite"
).with_structured_output(QuestionList)
# llm = ChatOllama(model="deepseek-r1").with_structured_output(QuestionList)
prompt = ChatPromptTemplate.from_messages([
    ("system","You are a AWS Solutions Architect. based on following service name, "
    "provide a relevant questions which can be asked to the customer to understand their requirements better."
    "Come up with a list of questions that will help you gather the necessary information to design a solution for the customer."
    "we can help them get best configuration and pricing for the service."
    "You can create a list of 10-20 questions that will help you gather the necessary information & create questions and have deterministic answers as options."
    "striclty stick to service name and data provided in the input."),
    ("human","AWS {service_name}  service. Here is the Pricing & Configuration JSON data: {data}")])

agent = prompt | llm

result = agent.invoke(input={'service_name': service_name, 'data': data})

for question in result.questions:
    print(f"Question: {question.question}")
    for i, option in enumerate(question.options, start=1):
        print(f"Option {i}: {option}")
    print()  # Print a newline for better readability
```
